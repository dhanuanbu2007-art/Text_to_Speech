```python
import gradio as gr
from gtts import gTTS
import tempfile

def text_to_speech(text, language):

    if not text.strip():
        return None

    try:
        speech = gTTS(text=text, lang=language)

        temp_file = tempfile.NamedTemporaryFile(
            suffix=".mp3",
            delete=False
        )

        speech.save(temp_file.name)

        return temp_file.name

    except Exception as e:
        print("Error:", e)
        return None


demo = gr.Interface(
    fn=text_to_speech,

    inputs=[
        gr.Textbox(
            lines=5,
            label="Enter Your Text",
            placeholder="Type something here..."
        ),

        gr.Dropdown(
            choices=[
                ("English", "en"),
                ("Tamil", "ta"),
                ("Hindi", "hi")
            ],
            value="en",
            label="Select Language"
        )
    ],

    outputs=gr.Audio(
        type="filepath",
        label="Generated Speech"
    ),

    title="🔊 Text-to-Speech Converter",

    description="Convert written text into spoken audio in different languages.",

    examples=[
        ["Hello, welcome to our Text-to-Speech application.", "en"],
        ["வணக்கம்! எப்படி இருக்கிறீர்கள்?", "ta"],
        ["नमस्ते! आप कैसे हैं?", "hi"]
    ]
)

demo.launch(share=True)
```
